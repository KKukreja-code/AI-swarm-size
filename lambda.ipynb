```python
import numpy as np
from scipy.special import logit, expit
from scipy.stats import t, norm
import statsmodels.api as sm
```

```python
# Browse
x = np.array([0.88, 2.53, 4.33, 5.63, 7.99, 1, 1.83, 2.99, 4.07, 6.58, 0.89, 1.63, 2.79, 4.02, 6.69])
N = np.array([1, 1, 1, 1, 1, 4, 4, 4, 4, 4, 16, 16, 16, 16, 16])
score = np.array([69.04, 84.6, 88.23, 89.42, 90.84, 81.75, 89.18, 90.76, 91.75, 92.18, 86.41, 91.07, 92.18, 93.36, 93.29])
score /= 100
# X = np.column_stack((x, N, x * N))
X = np.column_stack((np.log(x), np.log(N)))

X = sm.add_constant(X)

y = logit(score)

model = sm.OLS(y, X).fit()

print(model.summary())
```

```python
# 95% confidence interval for lambda_browse
b1, b2 = model.params[1:3]
lam_browse = b2 / b1

cov = np.asarray(model.cov_params())[1:3, 1:3]
grad = np.array([-b2 / b1**2, 1 / b1])
var_lam = grad @ cov @ grad
se_browse = np.sqrt(var_lam)

t_crit = t.ppf(0.975, df=model.df_resid)
ci_lower = lam_browse - t_crit * se_browse
ci_upper = lam_browse + t_crit * se_browse

print(f"Lambda = {lam_browse}")
print(f"Standard error = {se_browse}")
print(f"95% CI = [{ci_lower}, {ci_upper}]")
```

```python
# SEC-Bench

x = np.array([1.58,4.46,7.42,14.25,20.57,1.55,3.19,5,7.43,11.85,1.61,2.87,3.84,6.15,7.97])
N = np.array([1, 1, 1, 1, 1, 4, 4, 4, 4, 4, 16, 16, 16, 16, 16])
score = np.array([17.1,37.3,53.1,64.1,71.5,20.4,39.0,50.1,65.3,74.3,22.4,45.6,61.5,66.7,76.2])
score /= 100
# X = np.column_stack((x, N, x * N))
X = np.column_stack((np.log(x), np.log(N)))

X = sm.add_constant(X)

y = logit(score)

model = sm.OLS(y, X).fit()

print(model.summary())
```

```python
b1, b2 = model.params[1:3]
lam_SEC = b2 / b1

cov = np.asarray(model.cov_params())[1:3, 1:3]
grad = np.array([-b2 / b1**2, 1 / b1])
var_lam = grad @ cov @ grad
se_SEC = np.sqrt(var_lam)

t_crit = t.ppf(0.975, df=model.df_resid)
ci_lower = lam_SEC - t_crit * se_SEC
ci_upper = lam_SEC + t_crit * se_SEC

print(f"Lambda = {lam_SEC}")
print(f"Standard error = {se_SEC}")
print(f"95% CI = [{ci_lower}, {ci_upper}]")
```

```python
# Null hypothesis: lam_browse <= lam_SEC
diff = lam_browse - lam_SEC
se_diff = np.sqrt(se_browse**2 + se_SEC**2)
z = diff / se_diff

p_value = norm.sf(z) # complement of cdf

print(f"Difference in lambda: {diff}")
print(f"Standard error: {se_diff}")
print(f"One-sided p-value: {p_value}")
```
